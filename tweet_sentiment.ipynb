```python
# Importing Word2Vec (Gensim)
import gensim.downloader as api

# Loading pretrained Word2Vec
word2vec_model = api.load("word2vec-google-news-300")

print("Word2Vec model loaded successfully!")
```

```python
# Importing BERT (HuggingFace)
from transformers import BertTokenizer, BertModel

# Loading BERT tokenizer and model
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
bert_model = BertModel.from_pretrained("bert-base-uncased")

print("BERT model loaded successfully!")
```

# Embeddings of 5 adjectives using Word2Vec and BERT

```python
words = ["beautiful", "small", "blue", "pragmatic", "meticulous"]
```

```python
# Word2Vec Embeddings
print("Word2Vec Embeddings:\n")
for word in words:
    if word in word2vec_model:
        print(f"{word} -> {word2vec_model[word]}")
    else:
        print(f"{word} not in vocabulary")
```

```python
# BERT Embeddings
import torch
print("\nBERT Embeddings:\n")
def get_bert_embedding(word):
    inputs = tokenizer(word, return_tensors='pt')
    with torch.no_grad():
        outputs = bert_model(**inputs)
    return outputs.last_hidden_state[0][1]  # actual word embedding
for word in words:
    emb = get_bert_embedding(word)
    print(f"{word} -> {emb}")
```

# Top 5 most similar words for those adjectives using Word2Vec and BERT

```python
# Word2Vec Similar Words
print("Word2Vec - Top 5 Similar Words:\n")
for word in words:
    if word in word2vec_model:
        print(f"{word} ->")
        similar_words = word2vec_model.most_similar(word, topn=5)
        for w, score in similar_words:
            print(f"   {w} ({score:.4f})")
    else:
        print(f"{word} not in vocabulary")
    print()
```

```python
# BERT Similar Words (using Cosine Similarity)
from sklearn.metrics.pairwise import cosine_similarity
print("\nBERT - Top 5 Similar Words:\n")

# Comparison vocabulary
vocab = [
    # beautiful
    "pretty", "gorgeous", "lovely", "attractive", "stunning",
    # small
    "tiny", "little", "mini", "compact", "short",
    # blue
    "red", "green", "yellow", "dark", "bright", "purple",
    # pragmatic
    "practical", "realistic", "logical", "rational", "sensible",
    # meticulous
    "careful", "precise", "detailed", "thorough", "accurate",
    # some distractors
    "ugly", "huge", "careless", "irrational", "messy"
]
for word in words:
    word_emb = get_bert_embedding(word).reshape(1, -1)
    similarities = []
    for v in vocab:
        v_emb = get_bert_embedding(v).reshape(1, -1)
        sim = cosine_similarity(word_emb, v_emb)[0][0]
        similarities.append((v, sim))
    similarities.sort(key=lambda x: x[1], reverse=True)
    print(f"{word} -> {similarities[:5]}")
```

```python
import pandas as pd
import numpy as np
import torch
from sklearn.neural_network import MLPClassifier
import gensim.downloader as api
from transformers import BertTokenizer, BertModel
```

```python
train_df = pd.read_csv("train.csv", encoding="latin-1")
train_df = train_df[["text", "sentiment"]]

test_df = pd.read_csv("test.csv", encoding="latin-1")
test_df = test_df[["text", "sentiment"]]
```

## Data Cleaning and Preparation

```python
train_df = train_df.dropna(subset=["text", "sentiment"])
test_df = test_df.dropna(subset=["text", "sentiment"])
test_df = test_df.drop_duplicates()
```

```python
# Reassigning fresh row numbers
train_df = train_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)
```

```python
train_df["sentiment"] = train_df["sentiment"].map({
    "negative": 0,
    "neutral": 1,
    "positive": 2
})

X_train = train_df["text"].astype(str)
y_train = train_df["sentiment"].values



test_df["sentiment"] = test_df["sentiment"].map({
    "negative": 0,
    "neutral": 1,
    "positive": 2
})

X_test = test_df["text"].astype(str)
y_test = test_df["sentiment"].values
```

```python
print("NaN in y_test:", np.isnan(y_test).sum())
```

## 1. Word2Vec-based Neural Model

```python
word2vec_model = api.load("word2vec-google-news-300")
```

```python
# Converting a tweet into a vector
def tweet_to_vec(tweet):
    words = tweet.split()
    vecs = [word2vec_model[w] for w in words if w in word2vec_model]
    return np.mean(vecs, axis=0) if vecs else np.zeros(300)

X_train_w2v = np.array([tweet_to_vec(t) for t in X_train])
X_test_w2v = np.array([tweet_to_vec(t) for t in X_test])
```

```python
# Multilayer Perceptron classifier with one hidden layer with 128 neurons
model_w2v = MLPClassifier(hidden_layer_sizes=(128,), max_iter=300)
model_w2v.fit(X_train_w2v, y_train)
```

## 2. BERT-based Neural Model

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu") # Use GPU if available, otherwise CPU
```

```python
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
bert_model = BertModel.from_pretrained("bert-base-uncased").to(device)
bert_model.eval()
```

```python
# Converting text to BERT embeddings
def get_bert_embeddings(texts, batch_size=32):
    embeddings = []

    for i in range(0, len(texts), batch_size):
        batch = texts[i:i+batch_size].tolist()

        inputs = tokenizer(
            batch,
            return_tensors="pt",
            truncation=True,
            padding=True,
            max_length=64
        )

        inputs = {k: v.to(device) for k, v in inputs.items()}

        with torch.no_grad():
            outputs = bert_model(**inputs)

        cls_vec = outputs.last_hidden_state[:, 0, :].cpu().numpy()
        embeddings.extend(cls_vec)

    return np.array(embeddings)

X_train_bert = get_bert_embeddings(X_train)
X_test_bert = get_bert_embeddings(X_test)
```

```python
model_bert = MLPClassifier(hidden_layer_sizes=(128,), max_iter=300)
model_bert.fit(X_train_bert, y_train)
```

# Comparing both models using F1-score on test data

```python
from sklearn.metrics import f1_score
```

```python
# Word2Vec model evaluation
y_pred_w2v = model_w2v.predict(X_test_w2v)
f1_w2v = f1_score(y_test, y_pred_w2v, average='weighted')
print("Word2Vec Model F1-score:", f1_w2v)
```

```python
# BERT model evaluation
y_pred_bert = model_bert.predict(X_test_bert)
f1_bert = f1_score(y_test, y_pred_bert, average='weighted')
print("BERT Model F1-score:", f1_bert)
```

```python
# Comparing models
if f1_bert > f1_w2v:
    print("BERT model performs better.")
elif f1_w2v > f1_bert:
    print("Word2Vec model performs better.")
else:
    print("Both models perform equally.")
```
